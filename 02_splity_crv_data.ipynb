## Datasets

### Dataset: NeloreCRV

Dados de todos touros da raça Nelore presentes na plataforma CRV

```python
csv_path = 'Dados/CSV/NeloreCRV.csv'
df = pd.read_csv(csv_path)
print(f"Dimensões do dataset: {df.shape[0]} touros x {df.shape[1]} variáveis")
df.head()
```

```python
# Revisão v11 (out/2026): a base bruta passa a ser a do orientador (repositório nelore-hedonic-pricing, coleta de
#   outubro de 2026), com 229 touros que têm genealogia e preço da dose convencional. Os valores são os mesmos da nossa
#   coleta; mudam só os nomes das colunas gerais (em caixa alta) e a grafia de quatro índices. Esta célula traz os
#   nomes de volta ao padrão do notebook 01, usado pelos notebooks seguintes.
nomes_gerais = {
    'PRECO_CONV': 'preco_convencional', 'DNASC': 'data_nascimento', 'SIST_CRIACAO': 'sistema_criacao',
    'IDADE': 'idade_medicao', 'PESO': 'peso_kg', 'CIRC_ESC': 'circunferencia_escrotal_cm',
    'COMP_CORP': 'comprimento_corporal_cm', 'ALT_GAR': 'altura_garupa_cm', 'LARG_GAR': 'largura_garupa_cm',
    'COMP_GAR': 'comprimento_garupa_cm', 'PROF_UBERE': 'profundidade_torax_cm', 'PER_TOR': 'perimetro_toracico_cm'
}
indices = {'_MGTE_': '_MGTe_', '_IABCZ_': '_iABCZ_', '_PEIP_': '_PEip_', '_PEI_': '_PEi_'}
prefixos = ('ANCP_', 'PMGZ_', 'GENEPLUS_', 'DELTAGEN_', 'ALIANÇA_', 'CIA_', 'CFM_', 'IZ_', 'PAINT_', 'QUALITAS_', 'NELORE_')

def padroniza(c):
    if c in nomes_gerais:
        return nomes_gerais[c]
    if not c.startswith(prefixos):
        return c.lower()
    for antigo, novo in indices.items():
        c = c.replace(antigo, novo)
    return c

df.columns = [padroniza(c) for c in df.columns]
print(f"Dimensões do dataset: {df.shape[0]} touros x {df.shape[1]} variáveis")
print(df['raca'].value_counts().to_string())
```

### Construção do Dataset: NeloreCRV_Geral

Dados de registro, genealogia e fenotípicos de todos os touros da raça Nelore presentes na plataforma CRV

```python
cols_gerais = [
    'crv_code', 'nome_oficial', 'nome_guerra', 'raca', 'rgd', 'data_nascimento',
    'preco_convencional', 'preco_sexado',
    'peso_kg', 'circunferencia_escrotal_cm', 'altura_garupa_cm',
    'comprimento_corporal_cm', 'largura_garupa_cm', 'comprimento_garupa_cm',
    'profundidade_torax_cm', 'perimetro_toracico_cm',
    'criador', 'proprietario', 'linhagem', 'pai', 'mae', 
    'avo_paterno', 'avo_materno', 'selos', 'sistema_criacao', 'url_loja'
]

# Filtrar colunas existentes no dataset
cols_gerais_existentes = [c for c in cols_gerais if c in df.columns]
df_geral = df[cols_gerais_existentes].copy()

arquivo_geral = 'Dados/CSV/NeloreCRV_Dados_Gerais.csv'
df_geral.to_csv(arquivo_geral, index=False, encoding='utf-8-sig')

print("=" * 75)
print("1. ARQUIVO DE DADOS GERAIS GERADO COM SUCESSO:")
print(f"   -> Arquivo: {arquivo_geral}")
print(f"   -> Registros: {len(df_geral)} touros | Colunas: {len(df_geral.columns)}")
print("=" * 75)
```

##### Seleçao de colunas

```python
base_cols_reduzida = [
    'crv_code', 'nome_oficial', 'raca', 'rgd', 'data_nascimento', 'preco_convencional', 'selos'
]

base_cols_reduzida = [c for c in base_cols_reduzida if c in df.columns]
```

#### Dados dos programas de melhoramento

```python
programas = {
    'ANCP': {
        'prefixo': 'ANCP_',
        'coluna_chave': 'ANCP_MGTe_DEP',
        'arquivo': 'Dados/CSV/NeloreCRV_ANCP.csv'
    },
    'PMGZ': {
        'prefixo': 'PMGZ_',
        'coluna_chave': 'PMGZ_iABCZ_DEP',
        'arquivo': 'Dados/CSV/NeloreCRV_PMGZ.csv'
    },
    'GENEPLUS': {
        'prefixo': 'GENEPLUS_',
        'coluna_chave': 'GENEPLUS_IQG_DEP',
        'arquivo': 'Dados/CSV/NeloreCRV_Geneplus.csv'
    },
    'DELTAGEN': {
        'prefixo': 'DELTAGEN_',
        'coluna_chave': 'DELTAGEN_INDF_DEP',
        'arquivo': 'Dados/CSV/NeloreCRV_Deltagen.csv'
    },
    'ALIANÇA': {
        'prefixo': 'ALIANÇA_',
        'coluna_chave': 'ALIANÇA_INDF_DEP',
        'arquivo': 'Dados/CSV/NeloreCRV_Alianca.csv'
    }
}
```

```python


print("\n2. ARQUIVOS ESPECÍFICOS POR PROGRAMA (BASE ENXUTA + RÉGUAS GENÉTICAS):")

for prog_nome, config in programas.items():
    prefixo = config['prefixo']
    col_chave = config['coluna_chave']
    arquivo_saida = config['arquivo']
    
    # Selecionar apenas as colunas que pertencem ao programa (ex: que começam com 'ANCP_')
    prog_cols = [c for c in df.columns if c.startswith(prefixo)]
    
    # Filtrar touros que participam do programa
    if col_chave in df.columns:
        sub_df = df[df[col_chave].notna()].copy()
    else:
        sub_df = df[df[prog_cols].notna().any(axis=1)].copy()
        
    # Manter colunas base reduzidas + colunas do programa
    cols_finais = base_cols_reduzida + prog_cols
    sub_df = sub_df[cols_finais]
    
    # Salvar em CSV
    sub_df.to_csv(arquivo_saida, index=False, encoding='utf-8-sig')
    
    com_preco = (sub_df['preco_convencional'].notna() & (sub_df['preco_convencional'] > 0)).sum()
    
    print(f"\n   [{prog_nome}]")
    print(f"   - Touros avaliados: {len(sub_df)}")
    print(f"   - Touros com Preço Convencional: {com_preco}")
    print(f"   - Colunas totais: {len(cols_finais)} (6 base + {len(prog_cols)} genéticas)")
    print(f"   - Salvo em: {arquivo_saida}")

print("\n" + "=" * 75)
print("[CONCLUÍDO] Todos os arquivos foram atualizados com sucesso!")
print("=" * 75)
```

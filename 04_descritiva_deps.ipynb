#### Análise da posição relativa da DEPs (*_TOP) por faixa

```python
# Definição das 4 Faixas de TOPs
def faixa_tops(val):
    if pd.isna(val): return np.nan
    if val <= 3.0:   return 'TOP 3%'
    if val <= 10.0:  return 'TOP 10%'
    if val <= 20.0:  return 'TOP 20%'
    return 'TOP > 20%'
```

```python
# Dicionário Oficial com TODAS as 18 características da ANCP
traits = {
    # Índice Geral ---
    'MGTe (Índice Geral)': 'MGTe_TOP',
    
    # Crescimento e Tamanho Corporal
    'P210 (Peso Desmama)': 'P210_TOP',
    'P365 (Peso ao Ano)': 'P365_TOP',
    'P450 (Peso Sobreano)': 'P450_TOP',
    'FRAME (Frame Corporal)': 'FRAME_TOP',
    # Revisão (out/2026): PAC é a Produtividade Acumulada (kg de bezerro desmamado por vaca por ano; ANCP),
    #   e não o peso da vaca.
    'PAC (Produtividade Acumulada)': 'PAC_TOP',

    # Reprodução, Fertilidade e Longevidade
    # Revisão (out/2026): rótulo conforme a ANCP: 3P é a Probabilidade de Parto Precoce.
    '3P (Prob. Parto Precoce)': '3P_TOP',
    'IPP (Idade 1º Parto)': 'IPP_TOP',
    'STAY (Stayability/Longevidade)': 'STAY_TOP',
    'PE450 (Perímetro Escrotal)': 'PE450_TOP',
    
    # Maternal e Parto
    'MP120 (Maternal 120d)': 'MP120_TOP',
    'MP210 (Maternal 210d)': 'MP210_TOP',
    'PN (Peso ao Nascer)': 'PN_TOP',
    'FPP (Facilidade Parto)': 'FPP_TOP',
    
    # Carcaça e Eficiência Alimentar
    'AOL (Musculatura Carcaça)': 'AOL_TOP',
    'ACAB (Gordura/Acabamento)': 'ACAB_TOP',
    'MARM (Marmoreio Carne)': 'MARM_TOP',
    'CAR (Consumo Residual/Eficiência)': 'CAR_TOP'
}
```

```python
# Processamento dos dados garantindo nomes de coluna padronizados
registros_completos = []

for nome, top in traits.items():
    if top in df_valid.columns:
        # Cria a série de faixas já com nome padronizado
        serie_faixa = df_valid[top].apply(faixa_tops)
        serie_faixa.name = 'faixas_tops'
        
        # Agrupamento estatístico do preço da dose (SEMEM)
        stats = df_valid.groupby(serie_faixa)['SEMEM'].agg(
            n_touros='count',
            media='mean',
            mediana='median',
            desvio='std',
            minimo='min',
            maximo='max'
        ).reset_index()
        
        stats['trait'] = nome
        registros_completos.append(stats)
```

```python
# Concatenar todos os resultados
df_analise_tops = pd.concat(registros_completos, ignore_index=True)
```

```python
# Reorganizar e arredondar colunas
cols_ordem = ['trait', 'faixas_tops', 'n_touros', 'media', 'mediana', 'desvio', 'minimo', 'maximo']
df_analise_tops = df_analise_tops[cols_ordem]

# Revisão v8 (out/2026): as médias deixam de ser arredondadas aqui. O ágio da célula seguinte era calculado sobre as médias já
#   arredondadas, o que mudava a 2ª casa (FRAME 11,5% em vez de 11,4%; P210 R$ 1,93 em vez de R$ 1,92).
#   O arredondamento passa a ser feito só na exibição.
```

```python
# 1. Definir a ordem exata das faixas
ordem_faixas = ['TOP 3%', 'TOP 10%', 'TOP 20%','TOP > 20%']

# 2. Converter para Categórico ordenado e ordenar o DataFrame
df_analise_tops['faixas_tops'] = pd.Categorical(
    df_analise_tops['faixas_tops'], 
    categories=ordem_faixas, 
    ordered=True
)

df_analise_tops = df_analise_tops.sort_values(by=['trait', 'faixas_tops'])

# 3. Reorganizar colunas
cols_ordem = ['trait', 'faixas_tops', 'n_touros', 'media', 'mediana', 'desvio', 'minimo', 'maximo']
df_analise_tops = df_analise_tops[cols_ordem]

# 4. Exibir com 2 casas (os valores guardados ficam sem arredondar; ver revisão acima)
display(df_analise_tops.round(2))
```

```python
# Tabela 1: Preço Médio por Faixa + Diferença (Ágio) da Super Elite vs Comercial
pivot_media = df_analise_tops.pivot_table(index='trait', columns='faixas_tops', values='media')
col_super = 'TOP 3%'
col_comercial = 'TOP > 20%'

if col_super in pivot_media.columns and col_comercial in pivot_media.columns:
    # 2. Calcula as colunas de ágio
    agio_reais = (pivot_media[col_super] - pivot_media[col_comercial]).round(2)
    agio_perc = (((pivot_media[col_super] / pivot_media[col_comercial]) - 1) * 100).round(1)

# Tabela 2.Ágio entre os TOP 3% e TOP > 20%
df_agio_extremos = pd.DataFrame({
        'Preço TOP 3% (R$)': pivot_media[col_super].round(2),
        'Preço TOP > 20% (R$)': pivot_media[col_comercial].round(2),
        'Ágio (R$)': agio_reais,
        'Ágio (%)': agio_perc
    })

# Tabela 3: Preço Mediano por Faixa
pivot_mediana = df_analise_tops.pivot_table(index='trait', columns='faixas_tops', values='mediana')

# Tabela 4: Número de Animais (N) por Faixa
pivot_n = df_analise_tops.pivot_table(index='trait', columns='faixas_tops', values='n_touros', fill_value=0).astype(int)



# Exibição formatada

print("=" * 95)
print("Estatísticas descritivas do valor da dose de semem agrupada por característica e classes TOPs")
print("=" * 95)
display(df_analise_tops.round(2))


print("=" * 95)
print("Ágios nos valores das doses de semem entre as classes TOP 3% e TOP > 20%")
print("=" * 95)
display(df_agio_extremos)


```

#### Gráfico de cascata (Waterfall Chart) 

Evidencia a agregação de valor a dose de semem a medida que há promoção de faixa de TOPs

##### Função: gera_graph_cascata

```python
def gera_graph_cascata(df, col_top, trait):

    # Calcular a média de preço em cada uma das 4 faixas
    serie_faixa = df[col_top].apply(faixa_tops)
    medias = df.groupby(serie_faixa)['SEMEM'].mean()
    
    # Ordem do degrau: Comercial -> Superior -> Elite -> Super Elite
    ordem_faixas = [
        'TOP > 20%',
        'TOP 20%',
        'TOP 10%',
        'TOP 3%'
    ]
    
    val_comercial = medias.get(ordem_faixas[0], 0)
    val_superior  = medias.get(ordem_faixas[1], 0)
    val_elite     = medias.get(ordem_faixas[2], 0)
    val_super     = medias.get(ordem_faixas[3], 0)
    
    # Diferenciais (degraus)
    degrau_superior = val_superior - val_comercial
    degrau_elite    = val_elite - val_superior
    degrau_super    = val_super - val_elite
    
    # Componentes do gráfico de cascata
    etapas = ['TOP > 20%', 'TOP 20%', 'TOP 10%', 'TOP 3%', 'TOTAL']
    valores_degrau = [val_comercial, degrau_superior, degrau_elite, degrau_super, val_super]
    
    # Altura base (bottom) de cada barra flutuante
    bottoms = [
        0,                                      # Base Comercial começa no zero
        val_comercial,                          # Degrau Superior apoia no Comercial
        val_comercial + degrau_superior,        # Degrau Elite apoia no Superior
        val_comercial + degrau_superior + degrau_elite, # Degrau Super Elite apoia na Elite
        0                                       # Barra final Super Elite vai do zero ao topo
    ]
    
    alturas = [
        val_comercial,
        degrau_superior,
        degrau_elite,
        degrau_super,
        val_super
    ]
    
    # Cores sofisticadas: Cinza (base), Azul/Verde (ganhos), Azul Escuro (total)
    cores = [
        '#7f8c8d',  # Base Comercial (Cinza)
        '#27ae60' if degrau_superior >= 0 else '#c0392b',  # Verde se positivo, Vermelho se negativo
        '#27ae60' if degrau_elite >= 0 else '#c0392b',
        '#2ecc71' if degrau_super >= 0 else '#c0392b',
        '#2c3e50'   # Total Super Elite (Azul Marinho)
    ]
    
    fig, ax = plt.subplots(figsize=(10, 6))
    
    barras = ax.bar(etapas, alturas, bottom=bottoms, color=cores, width=0.55, edgecolor='black', linewidth=0.8)
    
    # Adicionar rótulos numéricos nos degraus
    for i, (barra, val, bot) in enumerate(zip(barras, alturas, bottoms)):
        if i == 0 or i == 4:
            texto = f"R$ {val:.2f}"
            y_pos = val + 0.6
        else:
            sinal = "+" if val >= 0 else ""
            texto = f"{sinal}R$ {val:.2f}"
            y_pos = bot + val + (0.6 if val >= 0 else -1.2)
            
        ax.text(barra.get_x() + barra.get_width()/2, y_pos, texto, 
                ha='center', va='bottom', fontsize=11, fontweight='bold',
                color='#2c3e50' if (i==0 or i==4) else ('#1e8449' if val>=0 else '#922b21'))
    
    # Linhas de conexão pontilhadas entre as barras (efeito cascata clássico)
    for i in range(len(etapas) - 2):
        y_linha = bottoms[i+1] + alturas[i+1]
        ax.plot([i + 0.275, i + 1 - 0.275], [bottoms[i+1], bottoms[i+1]], color='gray', linestyle='--', linewidth=0.9)

    for spine in ax.spines.values():
        spine.set_visible(False)
    
    # Linha para a barra final
    ax.plot([2 + 0.275, 3 + 0.275], [val_super, val_super], color='gray', linestyle='--', linewidth=0.9)
    
    ax.set_title(f"Efeito da mudança de faixa TOP para a característica {trait} no preço da dose de semem", 
                 fontsize=14, fontweight='bold', loc = 'left', pad=15)
    ax.set_ylabel("Preço Médio da Dose (R$)", fontsize=12)
    ax.set_ylim(0, max(val_super, val_comercial, val_elite, val_superior) * 1.18)
    ax.yaxis.set_visible(False)

    ax.grid(False)
    
    plt.tight_layout()
    plt.show()
```

### Figuras do artigo (out/2026)

```python
# Revisão v8 (out/2026): Figura 1 do artigo refeita: largura fixa de 16 cm, fonte de 9 pt, coluna com o ágio e o N
#   de cada DEP (antes só algumas DEPs tinham o valor escrito e o N não aparecia), marcador cheio para as DEPs cujo IC
#   não inclui o zero (legível também em P&B). O estilo comum das figuras está em figuras_estilo.py.
from scipy import stats as st
from matplotlib.ticker import MultipleLocator
import figuras_estilo as fe
fe.estilo()

linhas = []
for nome, top in traits.items():
    faixa = df_valid[top].apply(faixa_tops)
    elite = df_valid.loc[faixa == 'TOP 3%', 'SEMEM']
    comercial = df_valid.loc[faixa == 'TOP > 20%', 'SEMEM']
    teste = st.ttest_ind(elite, comercial, equal_var=False)  # Welch: variâncias diferentes entre as classes
    ic = teste.confidence_interval(0.95)
    linhas.append({'DEP': top.replace('_TOP', ''), 'Ágio (R$)': elite.mean() - comercial.mean(),
                   'IC 95% inf.': ic.low, 'IC 95% sup.': ic.high, 'p (Welch)': teste.pvalue,
                   'N TOP 3%': len(elite), 'N TOP > 20%': len(comercial)})
df_agio_ic = pd.DataFrame(linhas).sort_values('Ágio (R$)').reset_index(drop=True)
print(f"Correção de Bonferroni para 18 comparações: p < {0.05 / 18:.4f}")
display(df_agio_ic.round(3))

fig, ax = plt.subplots(figsize=(fe.LARGURA, 4.9), layout='constrained')
ax.axvline(0, color=fe.TINTA_2, linewidth=0.8)
for i, r in df_agio_ic.iterrows():
    sig = r['IC 95% inf.'] > 0 or r['IC 95% sup.'] < 0
    cor = fe.AZUL if sig else fe.CINZA
    ax.plot([r['IC 95% inf.'], r['IC 95% sup.']], [i, i], color=cor, linewidth=1.6 if sig else 1.2, solid_capstyle='butt')
    ax.plot(r['Ágio (R$)'], i, 'o', markersize=5.5, color=cor, markerfacecolor=cor if sig else 'white',
            markeredgewidth=1.3, zorder=3)
    fe.coluna(ax, i, fe.br(r['Ágio (R$)'], 2, sinal=True), 40, color=fe.TINTA, fontweight='bold' if sig else 'normal')
    fe.coluna(ax, i, f"{r['N TOP 3%']:d} / {r['N TOP > 20%']:d}", 82, color=fe.TINTA_2)
fe.coluna(ax, len(df_agio_ic) - 0.1, 'Ágio (R$)', 40, color=fe.TINTA)
fe.coluna(ax, len(df_agio_ic) - 0.1, 'N', 82, color=fe.TINTA)
ax.set_yticks(range(len(df_agio_ic)), df_agio_ic['DEP'])
ax.set_ylim(-0.6, len(df_agio_ic) + 0.3)
ax.xaxis.set_major_locator(MultipleLocator(2))
ax.xaxis.set_major_formatter(fe.virgula(0))
ax.set_xlabel('Ágio no preço da dose (R$): Super Elite (TOP ≤ 3%) menos Comercial (TOP > 20%)')
ax.tick_params(axis='y', length=0)
ax.spines['left'].set_visible(False)
fig.get_layout_engine().set(rect=(0, 0, 0.79, 1))  # reserva a faixa da direita para as colunas de texto
fe.salvar(fig, 'fig_agio_deps_ic95.png')
plt.show()
```

```python
# Revisão v8 (out/2026): célula nova. Figura 2 do artigo: preço por classe de TOP para CAR, MGTe e AOL, com os touros
#   individuais, a média e o IC de 95%. Mostra que o ágio do CAR aparece só na classe Super Elite, enquanto MGTe e AOL
#   sobem a partir da classe Elite (base para a especificação SUPER_CAR do caderno 13).
import numpy as np
from scipy import stats as st
from matplotlib.ticker import MultipleLocator
import figuras_estilo as fe
fe.estilo()

CLASSES = [('TOP > 20%', 'Comercial\n> 20%'), ('TOP 20%', 'Superior\n11–20%'), ('TOP 10%', 'Elite\n4–10%'),
           ('TOP 3%', 'Super Elite\n≤ 3%')]
caracts = [('CAR_TOP', 'CAR'), ('MGTe_TOP', 'MGTe'), ('AOL_TOP', 'AOL')]
fig, axs = plt.subplots(1, 3, figsize=(fe.LARGURA, 2.9), sharey=True, layout='constrained')
fig.get_layout_engine().set(wspace=0.04)
resumo = []
for ax, (col, nome) in zip(axs, caracts):
    faixa = df_valid[col].apply(faixa_tops)
    ax.grid(axis='y', color=fe.GRADE, linewidth=0.6)
    ax.grid(axis='x', visible=False)
    medias = []
    for j, (cl, _) in enumerate(CLASSES):
        v = df_valid.loc[faixa == cl, 'SEMEM']
        x = j + np.random.default_rng(j).uniform(-0.16, 0.16, len(v))  # dispersão fixa: patamares de preço lado a lado
        ax.scatter(x, v, s=7, color=fe.CINZA_PONTO, alpha=0.55, linewidths=0, zorder=2)
        m = v.mean()
        h = st.t.ppf(0.975, len(v) - 1) * v.std(ddof=1) / np.sqrt(len(v))
        ax.errorbar(j, m, yerr=h, fmt='none', ecolor=fe.AZUL, elinewidth=1.4, capsize=3, zorder=3)
        ax.plot(j, m, 'o', markersize=5.5, color=fe.AZUL, markeredgecolor='white', markeredgewidth=0.8, zorder=4)
        medias.append(m)
        resumo.append({'DEP': nome, 'Classe': cl, 'N': len(v), 'Média (R$)': round(m, 2),
                       'IC 95% inf.': round(m - h, 2), 'IC 95% sup.': round(m + h, 2)})
        ax.annotate(f'n = {len(v)}', (j, 0), xycoords=('data', 'axes fraction'), xytext=(0, 2),
                    textcoords='offset points', ha='center', va='bottom', fontsize=7.5, color=fe.TINTA_2)
    ax.plot(range(4), medias, color=fe.AZUL, linewidth=0.8, alpha=0.6, zorder=3)
    ax.set_xticks(range(4), [c[1] for c in CLASSES], fontsize=7.5)
    ax.set_xlim(-0.55, 3.55)
    ax.set_title(nome, fontsize=9.5, fontweight='bold')
    ax.tick_params(axis='x', length=0)
axs[0].set_ylabel('Preço da dose (R$)')
axs[0].set_ylim(12, 62)
for ax in axs[1:]:
    ax.tick_params(axis='y', length=0)
axs[0].yaxis.set_major_locator(MultipleLocator(10))
axs[0].yaxis.set_major_formatter(fe.virgula(0))
display(pd.DataFrame(resumo))
fe.salvar(fig, 'fig_preco_classes_top.png')
plt.show()
```
